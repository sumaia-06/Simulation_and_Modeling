###Task-1 Scalar &amp; Array Operations (Redo Section with Your Own Values):  
● Run the given code with scalar = 3.7 and a new array of at least 6 elements.  
● Show outputs of all built-in mathematical functions (sin, cos, tan, etc.).

```python
# Sample scalar and array inputs
scalar = 3.7
arr = np.array([6.6, 2.5, 8.3, 4, 5, 6])

# Trigonometric functions
sin_values = np.sin(arr)
cos_values = np.cos(arr)
tan_values = np.tan(arr)
asin_values = np.arcsin(np.clip(arr / np.max(arr), -1, 1))  # Normalized and Clipping for valid range [-1, 1]
acos_values = np.arccos(np.clip(arr / np.max(arr), -1, 1))
atan_values = np.arctan(arr)

# Exponential and logarithm
exp_values = np.exp(arr)
log_values = np.log(arr)  # Natural logarithm

# Absolute value and square root
abs_values = np.abs(arr)
sqrt_values = np.sqrt(arr)

# Remainder when divided by a scalar (e.g., 2)
rem_values = np.remainder(arr, 2)

# Rounding functions
round_values = np.round(arr)
floor_values = np.floor(arr)
ceil_values = np.ceil(arr)

# Display results
print("sin:", sin_values)
print("cos:", cos_values)
print("tan:", tan_values)
print("asin:", asin_values)
print("acos:", acos_values)
print("atan:", atan_values)
print("exp:", exp_values)
print("log:", log_values)
print("abs:", abs_values)
print("sqrt:", sqrt_values)
print("rem:", rem_values)
print("round:", round_values)
print("floor:", floor_values)
print("ceil:", ceil_values)
```

## Task-2. Create a 3×2 matrix with custom values. Perform and display results of:
    ○ Trigonometric functions
    ○ Exponential & logarithmic functions
    ○ Absolute value & square root  
    ○ Remainder when divided by 2
    ○ Rounding (round, floor, ceil)
Make sure to clip inputs where needed to avoid math errors (e.g., arcsin/arccos domain
issues).

```python
# Creating a 3x2 matrix
matrix = np.array([[4.4, 1.3], [6.2, 1.9], [3.0, 7.9]])

# Trigonometric functions
sin_values = np.sin(matrix)
cos_values = np.cos(matrix)
tan_values = np.tan(matrix)
asin_values = np.arcsin(np.clip(matrix / np.max(matrix), -1, 1))  # Clipping for valid range [-1, 1]
acos_values = np.arccos(np.clip(matrix / np.max(matrix), -1, 1))
atan_values = np.arctan(matrix)

# Exponential and logarithm
exp_values = np.exp(matrix)
log_values = np.log(matrix)  # Natural logarithm (log base e)

# Absolute value and square root
abs_values = np.abs(matrix)
sqrt_values = np.sqrt(matrix)

# Remainder when divided by a scalar (e.g., 2)
rem_values = np.remainder(matrix, 2)

# Rounding functions
round_values = np.round(matrix)
floor_values = np.floor(matrix)
ceil_values = np.ceil(matrix)

# Display results
print("Matrix:\n", matrix)
print("\nsin:\n", sin_values, 3)
print("\ncos:\n", cos_values, 3)
print("\ntan:\n", tan_values, 3)
print("\nasine:\n", asin_values, 3)
print("\nacosine:\n", acos_values, 3)
print("\natangent:\n", atan_values, 3)
print("\nexp:\n", exp_values, 3)
print("\nlog (natural):\n", log_values, 3)
print("\nabs:\n", abs_values, 3)
print("\nsqrt:\n", sqrt_values, 3)
print("\nrem (remainder when divided by 2):\n", rem_values, 3)
print("\nround:\n", round_values)
print("\nfloor:\n", floor_values)
print("\nceil:\n", ceil_values)
```

# Task-3. Your 2×3 matrix can have float values (e.g., 2.1, 4.7…).  
● Implement and display:

    ○ Max & Min (with indices)
    ○ Vector length
    ○ Sorted version (both row-wise descending and entire matrix descending)
    ○ Sum, Product
    ○ Median, Mean, Standard Deviation
Hint: Use np.sort() with axis=1 for row-wise sorting. Use .flatten() for entire matrix sorting.

```python
# Creating a 2x3 matrix
matrix = np.array([[44, 1.3, 6.2], [1.9, 3, 7.9]])

# 1. Maximum value and index of max element
max_value = np.max(matrix)
max_index = np.argmax(matrix)  # Index of max element

# 2. Minimum value and index of min element
min_value = np.min(matrix)
min_index = np.argmin(matrix)  # Index of min element

# 3. Vector length / Total size / Shape
matrix_size = matrix.size      # Total elements (6)
matrix_shape = matrix.shape    # Dimensions (2, 3)

# 4. Sorted version
row_sorted_desc = np.sort(matrix, axis=1)[:, ::-1]      # row-wise descending
matrix_sorted_desc = np.sort(matrix.flatten())[::-1]    # entire matrix descending

# 5. Sum of elements
sum_values = np.sum(matrix)

# 6. Product of elements
prod_values = np.prod(matrix)

# 7. Median value
median_value = np.median(matrix)

# 8. Mean value
mean_value = np.mean(matrix)

# 9. Standard deviation
std_dev = np.std(matrix)

# Display results
print("Original 2x3 Matrix:\n", matrix)
print("\nMax value:", max_value)
print("Index:", max_index)
print("Min value:", min_value)
print("Index:", min_index)
print("Total elements (size):", matrix_size)
print("Shape:", matrix_shape)
print("Row-wise Descending:\n", row_sorted_desc)
print("Entire Matrix Descending:", matrix_sorted_desc)
print("Sum of elements:", sum_values)
print("Product of elements:", round(prod_values, 4))
print("Median value:", median_value)
print("Mean value:", round(mean_value, 4))
print("Standard deviation:", round(std_dev, 4))
```
